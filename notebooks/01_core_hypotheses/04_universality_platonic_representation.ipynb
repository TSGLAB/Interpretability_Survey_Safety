## 2. Measuring Representation Similarity

### Centered Kernel Alignment (CKA)

CKA measures similarity between two representation matrices $X \in \mathbb{R}^{n \times p_1}$ and $Y \in \mathbb{R}^{n \times p_2}$:

$$\text{CKA}(X, Y) = \frac{\|Y^T X\|_F^2}{\|X^T X\|_F \|Y^T Y\|_F}$$

This is invariant to:
- Orthogonal transformations
- Isotropic scaling

```python
def center(X):
    """Center the columns of X."""
    return X - X.mean(dim=0)

def linear_CKA(X, Y):
    """
    Compute linear CKA between two representation matrices.
    
    Args:
        X: [n_samples, dim_x]
        Y: [n_samples, dim_y]
    """
    X = center(X)
    Y = center(Y)
    
    # Compute Frobenius norms
    XTX = X.T @ X
    YTY = Y.T @ Y
    YTX = Y.T @ X
    
    cka = (YTX.norm() ** 2) / (XTX.norm() * YTY.norm())
    return cka.item()

# Test with random data
X = torch.randn(100, 64)
Y = torch.randn(100, 128)
print(f"CKA between random matrices: {linear_CKA(X, Y):.4f}")
print(f"CKA of X with itself: {linear_CKA(X, X):.4f}")
```

1. **Training data matters:** Models trained on different domains may not converge
2. **Architecture constraints:** Different architectures may solve problems differently
3. **Scale effects:** Larger models may use qualitatively different strategies
4. **Fine-tuning:** Task-specific training can break universality
