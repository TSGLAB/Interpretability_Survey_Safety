### What the Jacobian lens measures

A **lens** reads a hidden state $h_\ell$ from the middle of the network as a distribution over
vocabulary tokens. The [logit lens](./01_logit_lens.ipynb) does this by applying the final
unembedding $W_U$ directly — which silently assumes every layer already writes in the *final layer's
basis*. Mid-stack that is false: representations drift as they pass through attention and MLP blocks,
and the readout collapses into noise.

The **Jacobian lens** ([Anthropic, 2026](https://transformer-circuits.pub/2026/workspace/)) first
transports $h_\ell$ into the final basis using the model's own **average linearized map** from layer
$\ell$ to the output, then unembeds:

$$\bar{J}_\ell \;=\; \mathbb{E}\!\left[\frac{\partial\, h_{\text{final},\,t'}}{\partial\, h_{\ell,\,t}}\right],
\qquad
\text{lens}_\ell(h) \;=\; \text{softmax}\!\Big(W_U\,\operatorname{norm}\big(\bar{J}_\ell\, h\big)\Big)$$

The expectation runs over source positions $t$, all later positions $t' \ge t$, and a corpus of
~1,000 prompts. $\bar{J}_\ell$ is a single $[d_\text{model}, d_\text{model}]$ matrix per layer,
**estimated by backpropagation — there is no training and no loss function.** Averaging over many
contexts is the crucial step: it keeps what an activation does *persistently* and discards what it
does in one specific forward pass.

| Lens | Transport $\bar{J}_\ell$ | How it is obtained | Character |
|---|---|---|---|
| Logit lens | $I$ (identity) | nothing — assumes the bases already agree | breaks down mid-stack |
| Tuned lens | $I + A_\ell$ | **learned** on an output-matching (KL) loss | correlational — can *skip ahead* to the output |
| Jacobian lens | $\mathbb{E}\!\left[\partial h_{\text{final}}/\partial h_\ell\right]$ | **measured** by backprop, no training | causal, first-order — recovers early / middle layers |

Read the last column top to bottom for the point of the method. The **tuned lens** *learns* its
correction by fitting a **correlational, output-matching (KL) objective** — it is optimized to
reproduce the model's final output distribution. A fitted map therefore carries its own **inductive
bias**: it reflects what best predicts the final tokens, not what the model's weights actually *do*
with the activation, so it can "skip ahead" to the output and paper over genuine mid-stack
computation. The **Jacobian lens learns nothing** — it *derives* its transport straight from the
model's own weights with calculus, a **causal, first-order** map. Its readout is what the model itself
would *linearly do* with that activation, which is why it stays faithful in the early and middle
layers and surfaces **verbalizable representations** — concepts the model is disposed to make sayable,
whether or not they ever reach the output.

### The verbalizable workspace

The paper's larger claim is that the high-$\bar{J}_\ell$ directions form a **global workspace**: in a
middle band of layers only ~10–25 such directions are meaningfully active at once, they are causally
load-bearing (swapping one changes the model's stated answer), and they can be read out as verbal
report even when the concept never appears in the output. Early and late layers lack these
properties — late layers simply align with the imminent next token. Crucially the workspace is
*small*: the J-space accounts for **less than ~10% of activation variance** at any layer, so most of
what the residual stream computes is not verbalizable this way. This notebook demonstrates the
*reading* tool; see the paper for the workspace experiments (modulation, selectivity, steering).

```python
import jlens

jlens.configure_logging()

MODEL_NAME = "Qwen/Qwen3.5-4B"
# MODEL_NAME = "Qwen/Qwen3.6-27B"

LENS_REPO = "neuronpedia/jacobian-lens"
LENS_REVISION = "qwen-n1000"
LENS_FILE = {
    "Qwen/Qwen3.5-4B": "qwen3.5-4b/jlens/Salesforce-wikitext/Qwen3.5-4B_jacobian_lens_n1000.pt",
    "Qwen/Qwen3.6-27B": "qwen3.6-27b/jlens/Salesforce-wikitext/Qwen3.6-27B_jacobian_lens_n1000.pt",
}[MODEL_NAME]
```

## 1. Load the model

`jlens.from_hf` wraps an already-loaded HuggingFace model into a `LensModel` interface.

```python
import torch
import transformers

hf_model = transformers.AutoModelForCausalLM.from_pretrained(
    MODEL_NAME, dtype=torch.bfloat16
).cuda()
tokenizer = transformers.AutoTokenizer.from_pretrained(MODEL_NAME)
model = jlens.from_hf(hf_model, tokenizer)
model
```

## 2. Load a pre-fitted lens

`JacobianLens.from_pretrained` pulls a `.pt` from the Hub (or a local path). The lens holds one
`[d_model, d_model]` matrix per layer.

```python
lens = jlens.JacobianLens.from_pretrained(
    LENS_REPO, filename=LENS_FILE, revision=LENS_REVISION
)
lens
```

## 3. Apply: J-lens vs logit lens

`lens.apply(model, prompt, positions=...)` runs one forward pass, transports each layer's residual
into the final-layer basis with `J_l`, and decodes through the model's own unembedding.
`use_jacobian=False` skips the transport — that's the vanilla logit lens.

Below: a two-hop factual question, read out at the boot token. The J-lens surfaces interpretable
tokens at layers where the logit lens is still noise.

```python
prompt = "Fact: The currency used in the country shaped like a boot is"
layers = [
    model.n_layers // 4,
    model.n_layers // 2,
    model.n_layers // 4 * 3,
    model.n_layers - 2,
]

jlens_logits, model_logits, _ = lens.apply(model, prompt, layers=layers, positions=[-2])
logit_lens, _, _ = lens.apply(
    model, prompt, layers=layers, positions=[-2], use_jacobian=False
)


def top5(logits):
    return [tokenizer.decode([t]) for t in logits.topk(5).indices]


for layer in layers:
    print(f"L{layer:>3} logit-lens: {top5(logit_lens[layer][0])}")
    print(f"L{layer:>3} J-lens:     {top5(jlens_logits[layer][0])}")
print(f"model:           {top5(model_logits[0])}")
```

At **L16** the J-lens surfaces ` Italy` — the unspoken middle hop (boot → *Italy* → euro) — where the
logit lens shows only noise; the model itself never emits "Italy" (its own next-token prediction is
` is`). This single readout is a snapshot: the interactive view below traces the concept across every
position and layer, and in §6 a lens we fit ourselves sharpens ` Italy` to **rank 1** here.

## 4. Render a slice page (inline)

`compute_slice` + `build_page` produce an interactive position × layer view of the lens's token
ranks (the `?` in the corner explains the controls). `mode="embed"` inlines everything so the page
is self-contained.

```python
import gzip
import json
import tempfile
import urllib.request
from pathlib import Path

from jlens.examples import EXAMPLES, resolve_prompt
from jlens.vis import build_page, compute_slice, notebook_iframe

# English gloss for Qwen's Chinese/Japanese/Korean vocab tokens (machine-generated,
# best-effort), shown next to the token in the page (alt_token=). The file ships in the
# jlens repo under assets/; pull it from there since pip install doesn't include it.
GLOSS_URL = "https://raw.githubusercontent.com/anthropics/jacobian-lens/main/assets/qwen_gloss.json.gz"
gloss_path = Path(tempfile.gettempdir()) / "qwen_gloss.json.gz"
if not gloss_path.exists():
    urllib.request.urlretrieve(GLOSS_URL, gloss_path)
gloss = {int(k): v for k, v in json.load(gzip.open(gloss_path)).items()}

example = next(e for e in EXAMPLES if e.slug == "multihop")
prompt = resolve_prompt(example, tokenizer)

slice_data = compute_slice(
    model,
    lens,
    prompt,
    layer_stride=2,
    # Empirically on Qwen, the interesting word tokens trail punctuation and
    # single-character tokens in the raw top-K; mask to word-like tokens only.
    mask_display=True,
)
page, _, _ = build_page(
    slice_data,
    prompt,
    title=example.section,
    description=example.description,
    alt_token=gloss,
)
notebook_iframe(page)
```

## 5. Render a slice page (served)

For longer prompts prefer `mode="fetch"`: `build_page` writes the data as sidecar files to `out_dir`
and the page fetches rank files lazily on pin, so it stays small regardless of how many tokens are
tracked.

```python
import os
import threading
from functools import partial
from http.server import HTTPServer, SimpleHTTPRequestHandler

example = next(e for e in EXAMPLES if e.slug == "ascii-face")
prompt = resolve_prompt(example, tokenizer)

slice_data = compute_slice(model, lens, prompt, mask_display=True)
serve_root = Path(tempfile.gettempdir()) / "jlens_slices"
out_dir = serve_root / example.slug
page, _, _ = build_page(
    slice_data,
    prompt,
    title=example.section,
    description=example.description,
    alt_token=gloss,
    mode="fetch",
    out_dir=out_dir,
)
(out_dir / "index.html").write_text(page)

if "_jlens_httpd" not in globals():
    _handler = partial(SimpleHTTPRequestHandler, directory=os.path.abspath(serve_root))
    _jlens_httpd = HTTPServer(("127.0.0.1", 0), _handler)
    threading.Thread(target=_jlens_httpd.serve_forever, daemon=True).start()
print(f"-> http://localhost:{_jlens_httpd.server_address[1]}/{example.slug}/")
```

## 6. Fitting

`fit(model, prompts)` computes `J_l` over the supplied prompts. 100 prompts is enough for a usable
lens; the released lenses use 1000. `dim_batch` is the memory knob — each prompt does
`ceil(d_model / dim_batch)` backward passes on a retained graph.

> We fit on **10 prompts** here to keep the runtime short (~1 min/prompt on one H200), into a
> separate `my_lens` so the pre-fitted 4B lens loaded above stays the one we use everywhere else.
> Everything else matches the walkthrough; bump `n_prompts` to 100+ for a lens you'd actually use.

```python
from jlens.examples import load_wikitext_prompts

prompts = load_wikitext_prompts(n_prompts=10)
tmp = Path(tempfile.gettempdir())
my_lens = jlens.fit(
    model, prompts, dim_batch=32, max_seq_len=128,
    checkpoint_path=str(tmp / "ckpt.pt"),
)
my_lens.save(str(tmp / "jacobian_lens.pt"))
my_lens
```

Our 10-prompt `my_lens` is a drop-in for the released 4B lens. Reading §3's prompt with both side by
side shows the small lens already recovers the effect — `J_l` is an average over source positions and
prompts, so it converges from very few prompts.

```python
prompt = "Fact: The currency used in the country shaped like a boot is"
logit_out, _, _ = lens.apply(model, prompt, layers=layers, positions=[-2], use_jacobian=False)
released_out, _, _ = lens.apply(model, prompt, layers=layers, positions=[-2])
mine_out, _, _ = my_lens.apply(model, prompt, layers=layers, positions=[-2])

for layer in layers:
    print(f"L{layer:>3} logit-lens:     {top5(logit_out[layer][0])}")
    print(f"L{layer:>3} J-lens (4B):    {top5(released_out[layer][0])}")
    print(f"L{layer:>3} J-lens (n=10):  {top5(mine_out[layer][0])}")
    print()
```

## 7. Extension: R-lens

Notice that below ~L6 the J-lens still returns morphological fragments (`ting`, `lessly`) rather than
concepts. **R-lens**
([Blank, Bhatia & Nanda, 2026](https://www.lesswrong.com/posts/nv8oedrnLXKRzNEL9/r-lens-making-j-lens-more-faithful-on-early-layers))
is a follow-up that makes the early layers faithful.

**The problem.** $\bar{J}_\ell$ is fit by backpropagating from the final-layer residual stream all
the way down to layer $\ell$. The further back the signal travels, the more linearization and
numerical error accumulates across layers — so early-layer readouts degrade into noise.

**The fix: LRP.** R-lens keeps the estimator identical but replaces the raw backward pass with
**Layer-wise Relevance Propagation** — it propagates *relevance coefficients* instead of plain
gradients, using three stop-gradient rules that keep error from blowing up through the network's
nonlinear components:

| Rule | What it detaches | Effect |
|---|---|---|
| **LN-rule** | the RMSNorm / LayerNorm denominator | treats the norm as a linear map, preventing relevance collapse |
| **Identity-rule** | the nonlinear factor of SiLU / GELU | makes the activation's backward pass a per-element linear map |
| **Half-rule** | half of the SwiGLU gate product | splits relevance evenly across the gate's two branches instead of double-counting |

These are *only* stop-gradients: the forward pass is unchanged and the extra cost is negligible.

The authors release matched J-lens / R-lens pairs at
[`camilablank/workspace-lenses`](https://huggingface.co/camilablank/workspace-lenses). Using their
J-lens as the control keeps the fitting configuration fixed, so only the backward rules differ.

```python
RLENS_REPO = "camilablank/workspace-lenses"

j25 = jlens.JacobianLens.from_pretrained(RLENS_REPO, filename="qwen3.5-4b/j-lens/lens.pt")
r25 = jlens.JacobianLens.from_pretrained(RLENS_REPO, filename="qwen3.5-4b/r-lens/lens.pt")
print("J-lens (control):", j25)
print("R-lens          :", r25)

prompt = "Fact: The currency used in the country shaped like a boot is"
early = list(range(0, 17, 2))

j25_logits, _, _ = j25.apply(model, prompt, layers=early, positions=[-2])
r25_logits, _, _ = r25.apply(model, prompt, layers=early, positions=[-2])

header = "J-lens (n=25)"
print(f"\n{'':>6}{header:<52}R-lens (n=25)")
print("-" * 104)
for L in early:
    print(f"L{L:>3}   {str(top5(j25_logits[L][0])):<52}{top5(r25_logits[L][0])}")
```

From **L2 onward** — where the J-lens is still emitting morphology (`ting`, `lessly`) — R-lens already
surfaces tokens about the prompt: ` boot`, `-boot`, ` boots`. The boot concept shows up about four
layers earlier under R-lens (L2) than under J-lens (L6). Two honest caveats on this particular run:
the very first layer (L0) is noisy for *both* lenses, and here the J-lens actually reaches `意大利`
(Italy) by L10 while R-lens gets there later — so on a single 4B prompt the win is early-layer
*coherence*, not a clean lead on the downstream concept.

That is expected. The authors report the advantage **grows with scale** and evaluate properly: with
**pass@10** — does the intermediate land in the top-10 readout — across five prompt categories
(multi-hop facts, multilingual reasoning, associations, typo correction, poetry) and dense + MoE
models up to **284B** parameters, where the gap is largest (e.g. *Japan* surfacing at layer 2 under
R-lens versus layer 14 under J-lens), with ablations showing R-lens's early-layer directions are more
causally important for the answer.

One caveat the authors stress: R-lens is a tool for **faithfully tracking a computation across
layers**, not for discovering wholly new concepts — most concepts first appear in later layers, where
plain J-lens already works.
